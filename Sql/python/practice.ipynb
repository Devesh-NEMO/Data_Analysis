```python
import duckdb
con = duckdb.connect("practice.db")
con.sql("CREATE TABLE IF NOT EXISTS employees (name VARCHAR, department VARCHAR, salary INTEGER)")
con.sql("SHOW TABLES").show()

con.sql("""
INSERT INTO employees VALUES
  ('Asha',   'Engineering', 90000),
  ('Ravi',   'Engineering', 85000),
  ('Nikhil', 'Engineering', 95000),
  ('Meera',  'Sales',       60000),
  ('Karan',  'Sales',       65000),
  ('Divya',  'Sales',       72000)
""")
con.sql("SELECT * FROM employees").show()

con.sql("""
SELECT department, AVG(salary) AS avg_salary
FROM employees
GROUP BY department
""").show()
```
